```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from prophet import Prophet 
import numpy as np
from sklearn.metrics import mean_squared_error

# --- 1. Define File and Column Names ---
HISTORICAL_FILE = 'stores_sales_forecasting.csv' 
DATE_COL = 'Order Date' 
TARGET_COLUMN = 'Sales' # Confirmed column name for this dataset


# --- 2. Load the Data with Fixes ---
print(f"Loading data from: {HISTORICAL_FILE}...")
try:
    # Use 'latin-1' encoding to resolve previous UnicodeDecodeError
    df_train = pd.read_csv(
        HISTORICAL_FILE,
        encoding='latin-1', 
        parse_dates=[DATE_COL] 
    )
    print("Data loaded successfully!")

except Exception as e:
    print(f"FATAL ERROR during loading: {e}")

print("\nFirst 5 Rows:")
display(df_train.head())
```

```python
# --- 1. Aggregate Data to Daily Time Series ---
# Select only the date and target columns
daily_sales = df_train[[DATE_COL, TARGET_COLUMN]].copy()
daily_sales = daily_sales.set_index(DATE_COL)

# Aggregate: sum the sales for every single day
daily_sales_ts = daily_sales.resample('D').sum()
print("Data aggregated to daily sales time series.")


# --- 2. Handle Zero Sales Imputation ---
zero_sales_count = (daily_sales_ts[TARGET_COLUMN] == 0).sum()

if zero_sales_count > 0:
    print(f"Warning: There are {zero_sales_count} days with zero sales. Imputing...")
    # Fill zeros with NA, then forward-fill (ffill)
    daily_sales_ts.replace(0, pd.NA, inplace=True)
    daily_sales_ts.fillna(method='ffill', inplace=True)
    print("Zero sales imputed using 'forward fill'.")
else:
    print("No days with zero sales detected.")
    
print("\nAggregated Time Series Head:")
display(daily_sales_ts.head())
```

```python
# --- 1. Visualize the Raw Time Series ---
plt.style.use('fivethirtyeight')
sns.set_palette("tab10") 

plt.figure(figsize=(15, 6))
plt.plot(daily_sales_ts.index, daily_sales_ts[TARGET_COLUMN], 
         label='Daily Sales', color='blue', linewidth=1.5)
plt.title('Daily Sales Over Time', fontsize=18)
plt.xlabel('Date', fontsize=14)
plt.ylabel(f'Total {TARGET_COLUMN}', fontsize=14) 
plt.grid(True)
plt.legend()
plt.show()


# --- 2. Train/Test Split ---
# Split date: Start of the last year in the dataset (2017)
split_date = '2017-01-01'

train_data = daily_sales_ts.loc[daily_sales_ts.index < split_date]
test_data = daily_sales_ts.loc[daily_sales_ts.index >= split_date]

print(f"\nData Split Complete:")
print(f"Training Set: {len(train_data)} days ({train_data.index.min()} to {train_data.index.max()})")
print(f"Testing Set: {len(test_data)} days ({test_data.index.min()} to {test_data.index.max()})")


# --- 3. Visualize the Split ---
plt.figure(figsize=(15, 6))
plt.plot(train_data.index, train_data[TARGET_COLUMN], label='Training Data', color='green')
plt.plot(test_data.index, test_data[TARGET_COLUMN], label='Testing Data', color='red')
plt.title('Train/Test Split of Daily Sales Time Series')
plt.legend()
plt.show()
```

```python
# --- 1. Prepare Data for Prophet ---
prophet_train_df = train_data.reset_index()
prophet_train_df.columns = ['ds', 'y'] # Prophet requires 'ds' (datestamp) and 'y' (target)
print("Prophet training data prepared.")


# --- 2. Instantiate and Fit the Model ---
model = Prophet(yearly_seasonality=True, changepoint_prior_scale=0.1)
print("\nTraining Prophet Model...")
model.fit(prophet_train_df)
print("Model training complete.")


# --- 3. Create Future Dates and Generate Forecast ---
periods_to_forecast = len(test_data)

future_dates = model.make_future_dataframe(
    periods=periods_to_forecast, 
    freq='D', 
    include_history=False
)

# Generate the forecast (this is your prediction!)
forecast = model.predict(future_dates)

print("\nForecast Generated. Key prediction columns:")
display(forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].head())
```

```python
# --- 1. Combine Actual and Predicted Data ---
actuals = test_data[TARGET_COLUMN].reset_index(drop=True)
predictions = forecast['yhat']

# --- 2. Calculate RMSE and Define predictions_clipped ---
# Clip predictions at zero (sales cannot be negative)
predictions_clipped = predictions.clip(lower=0) 

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(actuals, predictions_clipped))

print(f"\nModel Performance:")
print(f"Root Mean Squared Error (RMSE): {rmse:,.2f}")


# --- 3. Visualize Prediction vs. Actual ---
plt.figure(figsize=(15, 6))
plt.plot(test_data.index, actuals, label='Actual Sales (Test Data)', color='blue', linewidth=2)
plt.plot(test_data.index, predictions_clipped, label='Predicted Sales (yhat)', color='red', linestyle='--', linewidth=2)
plt.title(f'Prophet Forecast vs. Actual Sales (RMSE: {rmse:,.2f})')
plt.xlabel('Date')
plt.ylabel(f'Total {TARGET_COLUMN}')
plt.legend()
plt.show()
```

```python
# --- 1. Check the length of your predictions (should be 364 days)
num_predictions = len(predictions_clipped)
print(f"Total predictions to submit: {num_predictions}")

# --- 2. Recreate the 'id' column based on the typical sequence ---
# The IDs in the original submission file start at 3000888
start_id = 3000888 
end_id = start_id + num_predictions
ids = range(start_id, end_id)

# --- 3. Create the Submission DataFrame ---
submission_df = pd.DataFrame({
    'id': ids,
    'sales': predictions_clipped.values
})

# --- 4. Save the New Submission File ---
FINAL_SUBMISSION_FILE = 'final_sales_forecast.csv'
submission_df.to_csv(FINAL_SUBMISSION_FILE, index=False)

print("\nSUCCESS: Submission file created!")
print(f"File saved as: {FINAL_SUBMISSION_FILE}")
display(submission_df.head())
print("\n-- Download this file and upload it to Kaggle! --")
```
